**Distribution of sentiment scores**

```python
import ast

def convert_to_dict(text):
    return ast.literal_eval(text)

df['sentiment_scores'] = df['sentiment_scores'].apply(convert_to_dict)

def extract_sentiment_counts(row):
    return row['positive'], row['negative'], row['neutral']

df[['positive_count', 'negative_count', 'neutral_count']] = df['sentiment_scores'].apply(extract_sentiment_counts).apply(pd.Series)

total_positive = df['positive_count'].sum()
total_negative = df['negative_count'].sum()
total_neutral = df['neutral_count'].sum()

print("Total Positive:", total_positive)
print("Total Negative:", total_negative)
print("Total Neutral:", total_neutral)
```

**Correlation  between sentiment_scores and favorability**

```python
df['sentiment_scores_sum'] = df['sentiment_scores'].apply(lambda x: sum(x.values()))

correlation = df['sentiment_scores_sum'].corr(df['favorability'])

print("Correlation between sentiment_scores and favorability:", correlation)
```

##**Hypothesis 2**
**Are there genres which are  underrepresented on YouTubeMovies compared to user preferences?**


```python
import pandas as pd
from scipy.stats import chi2_contingency

# Load the data
df = pd.read_csv('/content/drive/MyDrive/Colab Datasets/movies_youtube_capstoneIIICL.csv')

# Create a contingency table
contingency_table = pd.crosstab(df['favorability'], df['genre'])

# Perform the chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Interpret the results
if p < 0.05:
    print('There is a statistically significant difference between the observed and expected frequencies.')
    print('This means that certain content categories or genres are underrepresented on YouTubeMovies compared to user preferences.')
else:
    print('There is no statistically significant difference between the observed and expected frequencies.')
```

```python
import pandas as pd
from scipy.stats import f_oneway

# Assuming 'df' is your DataFrame containing favorability ratings and genres
# Replace 'df' with the name of your DataFrame if it's different

# Example:
# df = pd.read_csv('your_dataset.csv')

# Perform ANOVA test
result = f_oneway(*[group['favorability'] for name, group in df.groupby('genre')])

# Print the ANOVA test result
print("ANOVA Test Result:")
print("F-statistic:", result.statistic)
print("p-value:", result.pvalue)

# Interpret the results based on the p-value
if result.pvalue < 0.05:
    print("There is a significant difference in favorability ratings among different genres.")
    # Perform post-hoc analysis if needed
else:
    print("There is no significant difference in favorability ratings among different genres.")
```
